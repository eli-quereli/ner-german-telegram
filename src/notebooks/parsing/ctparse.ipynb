```python
import pandas as pd
from ctparse import ctparse
from datetime import datetime
```

```python
file_name = '../temp/translations_2022_06_09-2.csv'
```

```python
df = pd.read_csv(file_name,  parse_dates=['date'])
```

```python
df.info()
```

```python
for i in df.index: 
    
    text = df.loc[i, 'text_DE']
    ts = df.loc[i, 'date']
    
    parsed_result = ctparse(text, ts)

    if parsed_result:
        print(type((type(parsed_result.resolution))))
        print((parsed_result.resolution))
        print('Text: ', text)
        print('Timestamp: ', ts)
        print('Extracted date: ', parsed_result, '\n')
```
