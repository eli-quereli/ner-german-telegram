## imports

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# https://matplotlib.org/stable/gallery/color/named_colors.html
```

```python
df = pd.read_csv('temp/demotermine_cleaned.csv', parse_dates=['date', 'edit_date'])
```

```python
len(df)
```

```python
df.info()
```

## date range of messages

```python
print("First message in dataset: ", min(df.date.values))
```

```python
print("Last message in dataset: ", max(df.date.values))
```

## message distribution over time

```python
df.info()
```

```python
%matplotlib inline
```

```python
plt.figure(figsize=(20,10))
sns.histplot(data=df.date, legend=True, bins=39)
plt.savefig('figures/message_count.png')
plt.show()
```

## Text length

```python
df.text_length.describe()
```

```python
plt.figure(figsize=(10,10))
sns.boxplot(data=df.text_length, palette="rocket_r")
plt.savefig('figures/text_length_box.png')
plt.show()
```

```python
plt.figure(figsize=(15,10))
sns.histplot(x=df.text_length, bins=100)
plt.savefig('figures/text_length_box.png')
plt.show()
```
