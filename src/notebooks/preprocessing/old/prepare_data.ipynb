## extract annotation results for all categories

```python
categories = ['conspiracy', 
              'antisemitism', 
              'conspiracy_stance', 
              'conspiracy_sub', 
              'antisemitism_sub', 
              'antisemitism_stance', 
              'others', 
              'technical']
```

```python
def extract_annotation_results_per_category(row, category):
    for i in range(len(row)):
        if row[i]['from_name'] == category:
            labels = row[i]['value']['choices']
            return ", ".join(labels)
```

```python
def extract_annotation_results(df, categories):
    res = df.copy()
    
    for category in categories:
        res[category] = res['annotation_results'].apply(lambda x: extract_annotation_results_per_category(x, category=category) if len(x)> 0 else None)
    
    return res
```

```python
data = extract_annotation_results(data, categories)
```

## clean text

```python
from unicodedata import normalize
import spacy
nlp = spacy.load("de_core_news_md")
```

```python
def is_ascii(s):
    """Check if the characters in string s are in ASCII, U+0-U+7F."""
    return len(s) == len(s.encode())
```

```python
def normalize_string_keep_umlauts(string_lowercase, umlauts = ['ä', 'ö', 'ü', 'ß']):
    """Normalize unicode strings, e.g. with special formatting, but keep umlauts"""
    out = []
    
    for c in string_lowercase: 
        if c in umlauts: 
            pass
        else: 
            c = normalize('NFKD', c)
            
        out.append(c)       
    return "".join([c for c in out])
```

```python
def preprocess_text(s, output_col):

    s = s.replace('\n', ' ').lower()

    # only normalize strings which are not already in ascii-format
    if not is_ascii(s):
        s = normalize_string_keep_umlauts(s)
        
    doc = nlp(s)
    if output_col == 'tokens':
        words = [token.text for token in doc if token.is_alpha]
    
    else:
        words = [token.text for token in doc]
        
    return " ".join(words)
```
