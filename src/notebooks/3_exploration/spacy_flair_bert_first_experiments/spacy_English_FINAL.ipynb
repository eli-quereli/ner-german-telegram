# Exploration of medium-size English-language spaCy model for NER in Telegram messages

Using 10 example texts selected from Telegram channels 'Demotermine' for first German-language NER experiments and for machine translation experiments. 

```python
import pandas as pd

file_name = '../../temp/translations_2022_06_09-2.csv'
data = pd.read_csv(file_name)
len(data)
```

```python
# Select cleaned texts as German-language example messages 
texts = data.text_EN
texts[0]
```

## Predict and display entities in texts

```python
import spacy

# Load English-language medium-size spaCy model
nlp = spacy.load("en_core_web_md")

def get_entities(texts): 
    for text in texts:
        
        doc = nlp(text)
        if doc.ents:
            print("Text: ", text, '\n')
            print("The following NER tags are found: ", [(ent.text, ent.label_) for ent in doc.ents], '\n\n')
```

```python
get_entities(texts)
```
