# Exploration of German-language Flair model for NER in Telegram messages

Using ten example texts selected from Telegram channels 'Demotermine' for first German-language NER experiments and for machine translation experiments. 

```python
import pandas as pd

file_name = '../../temp/translations_2022_06_09-2.csv'
data = pd.read_csv(file_name)
len(data)
```

```python
# Select cleaned texts as sample
texts = data.cleaned_text
texts[0]
```

## Predict and print entities in texts

```python
from flair.models import SequenceTagger
from flair.data import Sentence

# Load Flair Sequence Tagger
tagger = SequenceTagger.load("flair/ner-german")


def get_entities(texts):
    
    for text in texts: 
        # transform example text to Sentence
        sentence = Sentence(text)

        # use tagger to predict NER tags in sentence
        tagger.predict(sentence)

        # print sentence
        print(sentence, '\n')

        # print predicted NER spans
        print('The following NER tags are found:')

        # iterate over entities and print
        for entity in sentence.get_spans('ner'):
            print(entity)
        print('\n\n')
```

```python
get_entities(texts)
```
