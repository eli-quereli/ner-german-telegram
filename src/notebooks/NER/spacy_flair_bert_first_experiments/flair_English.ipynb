## imports

```python
import pandas as pd
from flair.data import Sentence
from flair.models import SequenceTagger
```

## load data

```python
file_name = '../temp/translations_2022_06_09-2.csv'
```

```python
data = pd.read_csv(file_name)
```

```python
df = data.copy()
```

## pick some sample texts

```python
samples = df.text_EN.sample(10, random_state=1)
```

```python
samples
```

## load tagger

```python
tagger_default = SequenceTagger.load("flair/ner-english")
```

```python
# tagger_lg = SequenceTagger.load("flair/ner-english-large")
```

```python
def get_ents(tagger, text):

    sentence = Sentence(text)

    # predict NER tags
    tagger.predict(sentence)

    # print sentence
    print(sentence)

    # print predicted NER spans
    print('The following NER tags are found:')
    
    # iterate over entities and print
    for entity in sentence.get_spans('ner'):
        print(entity)
```

```python
for text in samples:
    try:
        get_ents(tagger_default, text)
    except:
        pass
```
