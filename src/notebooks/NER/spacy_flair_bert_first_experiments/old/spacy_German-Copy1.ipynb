# Spacy

## load model

```python
nlp = spacy.load("de_core_news_md")
```

```python
def get_entities(texts): 

    for text in texts: 
        
        entities = {}
        
        try: 
            doc = nlp(text)
            
            if doc.ents:
                print([(ent.text, ent.label_) for ent in doc.ents])
    
        except: 
            pass
        
        
```

```python
def display_entities(texts):
    
    for text in texts: 
    
        try: 
            doc = nlp(text)
            sentence_spans = list(doc.sents)
            displacy.render(sentence_spans, style="ent")

        except:
            pass
```

```python
get_entities(samples)
```
